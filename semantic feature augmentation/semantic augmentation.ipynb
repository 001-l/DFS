```python
import pandas as pd
from openai import OpenAI
```

```python
file_path = "语义增强.xlsx"
data = pd.read_excel(file_path)
```

```python
# Specify the column names to extract
columns_to_extract = [
    'LN转移个数','腋窝淋巴结状态','PR', 'HER2+FISH', 'ki-67', 
    '手术前怀孕', '治疗后怀孕', '治疗后生产','目前月经情况'
]

# Extract the specified columns
df = data[columns_to_extract]
```

```python
client = OpenAI(
    api_key="******",# The key you created
    base_url="https://api.openai.com/v1" 
)
```

```python
message_template = (
    "【Field Value Definitions】\n"
    "- HER2+FISH:\n"
    "  0 = HER2 negative\n"
    "  1 = HER2 positive\n"
    "  2 = HER2 ++ and FISH negative\n"
    "  3 = HER2 ++ and FISH positive\n"
    "- PR: actual value\n"
    "- Ki-67: actual value\n"
    "- Number of Lymph Node Metastases: according to pathological findings\n"
    "- Axillary Lymph Node Status:\n"
    "  0 = No metastasis\n"
    "  1 = Metastasis\n"
    "- Preoperative pregnancy: number of times pregnant before surgery\n"
    "- Post-treatment pregnancy: Number of pregnancies after treatment\n"
    "- Post-treatment delivery: Number of deliveries after treatment\n"
    "- Current menstrual status:\n"
    "  0 = Postmenopausal for one year or more\n"
    "  1 = Regular menstruation in the past three months\n"
    "  2 = Irregular menstruation in the past three months\n"
    "  3 = No menstruation for more than three months\n"
    "  4 = No menstruation for more than six months\n"
    "  5 = No menstruation for more than twelve months\n"
    "  6 = Amenorrhea\n"
    "  7 = Missing value\n"
    "- Surgical Procedure:\n"
    "  1 = wide local excision + sentinel lymph node biopsy\n"
    "  2 = wide local excision + axillary lymph node dissection\n"
    "  3 = mastectomy + sentinel lymph node biopsy\n"
    "  4 = mastectomy + axillary lymph node dissection\n"
    "  5 = mastectomy + sentinel lymph node biopsy + breast reconstruction\n"
    "  6 = mastectomy + axillary lymph node dissection + breast reconstruction\n"
    "- Radiotherapy: 0 = no; 1 = yes\n"
    "- Use of goserelin (GnRH analogue) during chemotherapy: 0 = no; 1 = yes\n"
    "- Targeted Therapy (Herceptin or Herceptin + Perjeta): 0 = no; 1 = yes\n"
    "- Chemotherapy Regimen:\n"
    "  0 = no chemotherapy\n"
    "  1 = PA (Paclitaxel + Pirarubicin)\n"
    "  2 = TA (Docetaxel + Pirarubicin)\n"
    "  3 = AC-P (Pirarubicin + Cyclophosphamide-Paclitaxel)\n"
    "  4 = AC-T (Pirarubicin + Cyclophosphamide-Docetaxel)\n"
    "  5 = TCrb (Docetaxel/Nab-Paclitaxel + Carboplatin)\n"
    "  6 = TC (Docetaxel + Cyclophosphamide)\n"
    "  7 = AC (Pirarubicin + Cyclophosphamide)\n"
    "  8 = other regimens\n"
    "  9 = TAC\n"
    "- Endocrine Therapy Regimen:\n"
    "  0 = No endocrine therapy\n"
    "  1 = Toremifene Citrate\n"
    "  2 = Goserelin + AI (Letrozole/Anastrozole/Exemestane)\n"
    "  3 = Goserelin + Toremifene Citrate\n"
    "  4 = Oophorectomy + Toremifene Citrate/AI\n"
    
    "You are an experienced oncology clinician, familiar with breast cancer staging, molecular markers, and treatment options. "
    "Based only on the following clinical features, generate a concise, professional, and neutral clinical interpretation to serve as a text feature for a downstream machine-learning model predicting five-year disease-free survival. Use only the provided fields for your reasoning.\n"
    "Requirements:\n"
    "1. Explain clinical significance of each marker\n"
    "2. Analyze marker interactions (high-risk vs protective factors)\n"
    "3. Based on the patient's current characteristics, recommend the most appropriate treatment combination, including: Surgical Procedure, Radiotherapy, Use of goserelin (GnRH analogue) during chemotherapy, Targeted Therapy (Herceptin or Herceptin + Perjeta), Chemotherapy Regimen, and Endocrine Therapy Regimen\n"
    "4. Assess risk stratification and DFS prognosis\n\n"
    
    "Output must strictly follow the following format (see example below), with four paragraphs, each in the form of 'label: content', and the overall length about 200 words.\n"
    "Example Template:\n"
    "Example 1:\n"
    "Current status: Two positive lymph nodes with axillary involvement denote moderate invasiveness; PR 95% reflects strong hormone dependence and a protective effect; HER2 “++”/FISH– indicates no HER2-driven aggressiveness; Ki-67 0.1% signifies very low proliferation; regular menstruation.\n"
    "Interactions: High PR and minimal Ki-67 synergize to lower recurrence risk, while nodal metastases elevate it; absence of HER2 positivity precludes targeted therapy but reinforces reliance on endocrine blockade.\n"
    "Recommended treatment: Wide local excision + ALND preferred given nodal involvement; radiotherapy indicated; targeted therapy not applicable due to HER2-negative status; AC-T regimen suitable for moderate-risk case; Toremifene Citrate appropriate for strong hormone sensitivity; goserelin use during chemo unnecessary given menstrual status.\n"
    "Risk stratification: Moderate risk overall, balanced by favorable proliferation and hormone profiles; estimated 5-year DFS approximately 85–90%.\n"
    "Example 2:\n"
    "Current status: Zero positive lymph nodes and no axillary involvement denote low invasiveness; PR 70% reflects strong hormone dependence and protective effect; HER2 positivity indicates aggressive potential; Ki-67 5% signifies low proliferation; postmenopausal status.\n"
    "Interactions: High PR and low Ki-67 synergize to lower recurrence risk despite HER2-driven aggressiveness; node-negative status mitigate spread.\n"
    "Recommended treatment: Wide local excision + SLNB appropriate for node-negative disease; radiotherapy advised to enhance local control; given HER2 positivity, targeted therapy with Herceptin (± Perjeta) is strongly recommended; low proliferation and hormone responsiveness support use of Toremifene Citrate; chemotherapy with AC or TC may be considered; goserelin not required in postmenopausal status.\n"
    "Risk stratification: Intermediate risk overall due to untreated HER2 positivity balanced by favorable hormone and proliferation profiles; estimated 5-year DFS approximately 75–80%.\n\n"

    "Data follows:\n"
)
```

```python
# Function to construct the message string
def generate_message(row):
    data_str = "Number of Lymph Node Metastases: {}, Axillary Lymph Node Status: {}, PR: {}, HER2+FISH: {}, ki-67: {}, Preoperative pregnancy: {}, Post-treatment pregnancy: {}, Post-treatment delivery: {}, Current menstrual status: {}".format(
        *[row[field] for field in columns_to_extract]
    )
    full_message = message_template + data_str
    return full_message
```

```python
output_data = []
```

```python
# Iterate over each row of data and generate the input message
for index, row in df.iterrows():
    input_message = generate_message(row)
    print(f"Input Message for Row {index + 1}:\n{input_message}\n")

    # Call the large model to generate a response
    response = client.chat.completions.create(
        model="gpt-4o",
        #model="gpt-4.1",
        messages=[{"role": "user", "content": input_message}]
    )
    content = response.choices[0].message.content
    print(f"Response for Row {index + 1}:\n{content}\n")
    
    # Add the response to the output data
    output_data.append({"输入消息": input_message, "输出结果": content})
```

```python
# Convert the output data into a DataFrame
output_df = pd.DataFrame(output_data)
```

```python
# Print the output DataFrame
print(output_df)
```
